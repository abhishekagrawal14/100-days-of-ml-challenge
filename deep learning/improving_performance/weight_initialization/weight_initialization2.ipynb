```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import tensorflow
from tensorflow import keras
from keras.layers import Dense
from keras.models import Sequential
```

```python
from sklearn.datasets import make_moons
X , y = make_moons(n_samples = 500 , noise = 0.2)
```

```python
plt.scatter(X[:,0] , X[:,1] , c = y)
```

```python
model1 = Sequential()
model1.add(Dense(2, activation='relu', input_dim=2))
model1.add(Dense(1 , activation = 'sigmoid'))
```

```python
model1.summary()
```

```python
model1.compile(optimizer = 'Adam' , loss = 'binary_crossentropy' , metrics = ['accuracy'])
```

```python
model1.get_weights()
```

```python
initial_weights = model1.get_weights()
```

```python
model1.get_weights()[0].shape
```

```python
initial_weights[0] = np.ones(model1.get_weights()[0].shape)
initial_weights[1] = np.ones(model1.get_weights()[1].shape)
initial_weights[2] = np.ones(model1.get_weights()[2].shape)
initial_weights[3] = np.ones(model1.get_weights()[3].shape)
```

```python
model1.set_weights(initial_weights)
```

```python
model1.get_weights()
```

```python
history1 = model1.fit(X , y , epochs = 100 )
```

```python
model1.get_weights()
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X, y, clf=model1)
```
